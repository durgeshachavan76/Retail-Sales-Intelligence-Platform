```python
rfm_data = order_level[
    order_level['order_status'] == 'delivered'
].copy()

reference_date = (
    rfm_data['order_purchase_timestamp'].max()
    + pd.Timedelta(days=1)
)

print("Reference date:", reference_date)
```

```python
rfm = (
    rfm_data
    .groupby('customer_unique_id')
    .agg(
        last_purchase=('order_purchase_timestamp', 'max')
    )
    .reset_index()
)

rfm['recency'] = (
    reference_date - rfm['last_purchase']
).dt.total_seconds() / (24 * 60 * 60)

rfm.head()
```

```python
rfm['R_score'] = pd.qcut(
    rfm['recency'],
    5,
    labels=[5, 4, 3, 2, 1]
)

rfm['F_score'] = pd.qcut(
    rfm['frequency'].rank(method='first'),
    5,
    labels=[1, 2, 3, 4, 5]
)

rfm['M_score'] = pd.qcut(
    rfm['monetary'],
    5,
    labels=[1, 2, 3, 4, 5]
)
```

```python
rfm['RFM_score'] = (
    rfm['R_score'].astype(int).astype(str) +
    rfm['F_score'].astype(int).astype(str) +
    rfm['M_score'].astype(int).astype(str)
)

rfm.head()
```

```python
def frequency_score(x):
    if x == 1:
        return 1
    elif x == 2:
        return 2
    elif x == 3:
        return 3
    elif x == 4:
        return 4
    else:
        return 5

rfm['F_score'] = rfm['frequency'].apply(frequency_score)
```

```python
rfm['RFM_score'] = (
    rfm['R_score'].astype(int).astype(str) +
    rfm['F_score'].astype(int).astype(str) +
    rfm['M_score'].astype(int).astype(str)
)
```

```python
def rfm_segment(row):

    r = row['R_score']
    f = row['F_score']
    m = row['M_score']

    if r >= 4 and f >= 4 and m >= 4:
        return 'Champions'

    elif r >= 3 and f >= 2 and m >= 3:
        return 'Loyal Customers'

    elif r >= 4 and f >= 2:
        return 'Potential Loyalists'

    elif r >= 3 and m >= 4:
        return 'Big Spenders'

    elif r >= 4 and f == 1:
        return 'New Customers'

    elif r <= 2 and f >= 2:
        return 'At Risk'

    else:
        return 'Lost Customers'


rfm['segment'] = rfm.apply(rfm_segment, axis=1)
```
